```python
# Handling Missing Values
# take a copy of the original dataframe to avoid modifying it directly
df_clean = df.copy() 

numeric_columns = df_clean.select_dtypes(include=[np.number]).columns.tolist()
categorical_columns = df_clean.select_dtypes(include=['object', 'string']).columns.tolist()

for col in numeric_columns:
    # fill na = fillna() method is used to fill the missing values in the column
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())  # fill with median for numeric columns

for col in categorical_columns:
    df_clean[col] = df_clean[col].fillna(df_clean[col].mode())  # fill with mode for categorical columns

display (df_clean.isnull().sum())  # check if there are any remaining missing values
```

```python
duplicate_count = df_clean.duplicated().sum()
print(f"Number of duplicate rows: {duplicate_count}")

duplicates = df_clean[df_clean.duplicated(keep=False)]  # keep=False marks all duplicates as True
display(duplicates.head(20))  # display all duplicate rows
```

```python
# remove duplicates
before = len (df_clean)
df_clean = df_clean.drop_duplicates()
after = len (df_clean)

print(f"Before removing duplicates: {before}")
print(f"After removing duplicates: {after}")
```

```python
# inspecting the cleaned data and displaying the data types and examples  of data
for col in df_clean.columns:
    print (f"{col} -> df_clean['{col}'].data_type: {df_clean[col].dtype}, unique values: {df_clean[col].nunique()}")
    print (f"Example values: {df_clean[col].dropna().head(3).tolist()}")  # display first 3 non-null values of the column
```
