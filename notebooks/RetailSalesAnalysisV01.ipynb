```python
# Create a copy of the original dataset for data cleaning
df_cleann = df.copy()

# Display the original and copied dataset shapes
print("Original Dataset Shape:", df.shape)
print("Copied Dataset Shape:", df_cleann.shape)
```

```python
# Remove duplicate rows from the copied dataset
df_cleann = df_cleann.drop_duplicates()

# Display the updated dataset shape
print("Dataset Shape After Removing Duplicates:", df_cleann.shape)
```

```python
# Verify that the original dataset remains unchanged
print("Original Dataset Shape:", df.shape)
print("Cleaned Dataset Shape:", df_cleann.shape)
```

```python
# Count missing values in each column
missing_values = df_cleann.isnull().sum()

# Calculate the percentage of missing values
missing_percentage = (df_cleann.isnull().mean() * 100).round(2)

# Combine missing values and percentages into a summary
missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage
})

# Display columns with missing values only
display(missing_summary[missing_summary["Missing Values"] > 0])
```

```python
# calcluate KPIs for the dataset

## ==========================================
# 1. Calculate Key Business Performance Indicators (KPIs)
# ==========================================

# Calculate the total sales revenue generated across all orders
total_sales = df_cleann["Sales"].sum()

# Calculate the total profit generated from all sales
total_profit = df_cleann["Profit"].sum()

# Count the number of unique orders
total_orders = df_cleann["Order ID"].nunique()

# Calculate the total quantity of products sold
total_quantity = df_cleann["Quantity"].sum()

# Calculate the overall profit margin as a percentage
profit_margin = (total_profit / total_sales) * 100


# ==========================================
# 2. Create a KPI Summary Table
# ==========================================

# Organize the calculated KPIs into a DataFrame
kpi_summary = pd.DataFrame({
    "KPI": [
        "Total Sales",
        "Total Profit",
        "Total Orders",
        "Total Quantity Sold",
        "Profit Margin (%)"
    ],
    "Value": [
        total_sales,
        total_profit,
        total_orders,
        total_quantity,
        profit_margin
    ]
})


# ==========================================
# 3. Display the KPI Summary
# ==========================================

# Display the business performance indicators
display(kpi_summary)
```

```python
# ==========================================
# 1. Analyze Sales and Profit by Category
# ==========================================

# Group the data by product category and calculate total sales and profit

# groupby(): Groups the dataset based on one or more columns.

# agg(): Applies multiple aggregation functions to different columns
# within a single operation.
category_performance = (
    df_cleann.groupby("Category")
    .agg(
        # Calculate total sales, profit, and quantity for each category
        # "sum" is used to aggregate the values in each group.
    
        Total_Sales=("Sales", "sum"), 
        Total_Profit=("Profit", "sum"),
        Total_Quantity=("Quantity", "sum")
    )
    .sort_values(by="Total_Sales", ascending=False)
)

# Display category performance
display(category_performance)
```

```python
# ==========================================
# 2. Visualize Sales by Category
# ==========================================

plt.figure(figsize=(10, 5))

sns.barplot(
    data=category_performance.reset_index(),
    x="Category",
    y="Total_Sales"
)

plt.title("Total Sales by Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
# Rotate category labels to improve readability
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# 3. Visualize Profit by Category
# ==========================================

# Create a figure with a specified width and height
plt.figure(figsize=(10, 5))

# Create a bar chart to compare total profit across product categories
sns.barplot(
    data=category_performance.reset_index(),
    x="Category",
    y="Total_Profit"
)

# Add chart title and axis labels
plt.title("Total Profit by Category")
plt.xlabel("Product Category")
plt.ylabel("Total Profit")

# Rotate category labels for better readability
plt.xticks(rotation=45)

# Adjust spacing to prevent overlapping labels
plt.tight_layout()

# Display the chart
plt.show()
```

```python
# ==========================================
# 5. Analyze Sales by Geography
# ==========================================

# Group sales records by geographical region
# Calculate total sales and profit for each region
# Sort regions by total sales in descending order

geography_performance = (
    df_cleann.groupby("Geography")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum")
    )
    .sort_values(by="Total_Sales", ascending=False)
)

# Display geographical sales performance
display(geography_performance)
```

```python
# ==========================================
# 7. Visualize Top 10 Geographical Regions by Sales
# ==========================================

# Select the top 10 regions with the highest total sales
top_10_geographies = geography_performance.head(10).reset_index()

# Create a figure with a specified width and height
plt.figure(figsize=(10, 6))

# Create a bar chart to compare sales across the top 10 regions
sns.barplot(
    data=top_10_geographies,
    x="Geography",
    y="Total_Sales"
)

# Add chart title and axis labels
plt.title("Top 10 Geographical Regions by Sales")
plt.xlabel("Geography")
plt.ylabel("Total Sales")

# Rotate region labels for better readability
plt.xticks(rotation=45, ha="right")

# Adjust spacing to prevent overlapping labels
plt.tight_layout()

# Display the chart
plt.show()
```

```python
# ==========================================
# 8. Analyze Profit by Geography
# ==========================================

# Select the top 10 regions with the highest total profit
top_10_profit_geographies = (
    geography_performance
    .sort_values(by="Total_Profit", ascending=False)
    .head(10)
    .reset_index()
)

# Create a figure with a specified width and height
plt.figure(figsize=(10, 6))

# Create a bar chart to compare profit across the top 10 regions
sns.barplot(
    data=top_10_profit_geographies,
    x="Geography",
    y="Total_Profit"
)

# Add chart title and axis labels
plt.title("Top 10 Geographical Regions by Profit")
plt.xlabel("Geography")
plt.ylabel("Total Profit")

# Rotate region labels for better readability
plt.xticks(rotation=45, ha="right")

# Adjust spacing to prevent overlapping labels
plt.tight_layout()

# Display the chart
plt.show()
```
